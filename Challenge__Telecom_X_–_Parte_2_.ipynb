##  **Separación en Conjunto de Entrenamiento y Prueba**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
```

## **Tranformando la variable Explicativas**

***Idenficando variables***

```python
col_categoricas = X_train.select_dtypes(include='object').columns
col_numericas = X_train.select_dtypes(include=['int64', 'float64']).columns
```

***Creación del Preprocesador***

```python
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer

preprocesador = ColumnTransformer (
    transformers =[
        ('num', MinMaxScaler(), col_numericas),
        ('cat', OneHotEncoder(drop='if_binary', handle_unknown='ignore'), col_categoricas)
    ]
)
```

##Modelado Predictivo

##**Creación de modelos**

***Modelo 1: Logistic Regression***

1.   Elemento de la lista
2.   Elemento de la lista



```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipeline_log = Pipeline(steps=[
    ('preprocessing', preprocesador),
    ('model', LogisticRegression(max_iter=1000))
])

pipeline_log.fit(X_train, y_train)
```

***Modelo 2: Random Forest***

```python
from sklearn.ensemble import RandomForestClassifier

pipeline_rf = Pipeline(steps=[
    ('preprocessing', preprocesador),
    ('model', RandomForestClassifier(
        n_estimators=200,

        random_state=42
    ))
])

pipeline_rf.fit(X_train, y_train)
```

## **Evaluación de los modelos**

Primero hacemos la predición para ambos

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, accuracy_score

# Logistic
y_pred_log = pipeline_log.predict(X_test)
y_proba_log = pipeline_log.predict_proba(X_test)[:,1]

# Random Forest
y_pred_rf = pipeline_rf.predict(X_test)
y_proba_rf = pipeline_rf.predict_proba(X_test)[:,1]
```

**Métricas – Logistic Regression**

```python
print("=== Logistic Regression ===")
print("Accuracy:", accuracy_score(y_test, y_pred_log))
print("\nMatriz de Confusión:")
print(confusion_matrix(y_test, y_pred_log))
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred_log))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_log))
```

** Métricas – Random Forest**

```python
print("=== Random Forest ===")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("\nMatriz de Confusión:")
print(confusion_matrix(y_test, y_pred_rf))
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_rf))
```

Interpretación Estratégica

**Diagnóstico de Desempeño y Validación Cruzada**

El proceso de evaluación técnica permite determinar la viabilidad operativa de los modelos de clasificación. A continuación, se detallan los hallazgos basados en métricas de rendimiento y capacidad de generalización:

1. **Evaluación de Métricas de Clasificación**

Exactitud (Accuracy) vs. Línea Base: Ambos modelos presentan un rendimiento en el rango del 79–80%. Dado que la clase mayoritaria representa el 73% de la muestra, un clasificador trivial alcanzaría dicho valor por azar; por lo tanto, el incremento obtenido confirma que las variables seleccionadas poseen poder predictivo real.

Sensibilidad (Recall) de Clase Crítica: La Regresión Logística reporta un Recall de 0.54 frente al 0.51 de Random Forest. En un contexto de Churn, priorizamos la detección del 54% de cancelaciones, identificando a la Regresión Logística como el modelo con mayor utilidad práctica.

Capacidad Discriminativa (ROC-AUC): Con un AUC de 0.84, la Regresión Logística supera al 0.82 de Random Forest, demostrando una robustez superior para separar las clases en distintos umbrales de decisión.

2. **Análisis de Estabilidad y Generalización**

Regresión Logística (Modelo Estable): La varianza entre el set de entrenamiento (0.809) y prueba (0.797) es mínima (~1%), lo que evidencia una óptima capacidad de generalización y ausencia de sesgos significativos.

Random Forest (Sobreajuste Detectado): Se observa una brecha crítica del 21% entre entrenamiento (0.997) y prueba (0.787). Este fenómeno de Overfitting indica que el modelo ha memorizado el ruido de los datos de entrenamiento, probablemente debido a una profundidad excesiva del árbol o falta de regularización.

3. **Conclusión de Desempeño Final**

El diagnóstico descarta la presencia de Underfitting, dado que ambos modelos operan por encima de los umbrales de aceptación técnica (>0.80 AUC). No obstante, la Regresión Logística se consolida como la solución superior en esta etapa inicial debido a su estabilidad estructural, mejor balance entre precisión/recall y su probada capacidad para generalizar ante datos no vistos.

4. **Hoja de Ruta para Optimización (Next Steps)**

Para elevar el Recall actual (captura del 54%) y minimizar la pérdida del 46% de clientes en riesgo, las próximas iteraciones técnicas incluirán:

Balanceo de Carga: Implementación de class_weight='balanced' para compensar el desequilibrio de clases.

Ajuste de Umbral: Calibración del threshold de decisión más allá del estándar 0.5 para optimizar la sensibilidad.

Tuning de Hiperparámetros: Regularización en la Regresión Logística y poda (pruning) en el Random Forest para mitigar la varianza.
